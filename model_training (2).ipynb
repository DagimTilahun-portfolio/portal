```python
# Import the Random Forest classification algorithm from scikit-learn
from sklearn.ensemble import RandomForestClassifier

# Import accuracy_score to calculate the percentage of correct predictions
from sklearn.metrics import accuracy_score

# Import precision_score to measure how many predicted positive cases were actually positive
from sklearn.metrics import precision_score

# Import recall_score to measure how many actual positive cases were correctly detected
from sklearn.metrics import recall_score

# Import f1_score to calculate the balance between precision and recall
from sklearn.metrics import f1_score

# Import confusion_matrix to see the number of TP, TN, FP, and FN predictions
from sklearn.metrics import confusion_matrix
```

```python
# Create a Random Forest classifier with 200 decision trees
rf_model = RandomForestClassifier(

    # Use 200 trees to make the model more stable than using only a few trees
    n_estimators=200,

    # Set a fixed random seed so we get reproducible results every time
    random_state=2
)
```

```python
# Calculate the percentage of test predictions that are correct
test_accuracy = accuracy_score(Y_test, Y_test_prediction)

# Print the test accuracy
print("Test Accuracy:", test_accuracy)
```

```python
# Calculate precision using the actual test labels and predicted test labels
test_precision = precision_score(Y_test, Y_test_prediction)

# Print the test precision
print("Test Precision:", test_precision)
```

```python
# Calculate recall to see how many actual heart-disease cases were detected
test_recall = recall_score(Y_test, Y_test_prediction)


print("Test Recall:", test_recall)
```

```python
# Calculate the F1 score, which balances precision and recall
test_f1 = f1_score(Y_test, Y_test_prediction)

print("Test F1 Score:", test_f1)
```

```python
# Print a heading to make the results easier to read
print("Random Forest Results")


print("Training Accuracy :", training_accuracy)

print("Test Accuracy     :", test_accuracy)

print("Test Precision    :", test_precision)

print("Test Recall       :", test_recall)

print("Test F1 Score     :", test_f1)
```

```python
# Import cross_val_score so we can evaluate the model using multiple train/validation splits
from sklearn.model_selection import cross_val_score

# Perform 5-fold cross-validation using the training dataset
cv_scores = cross_val_score(

    # Use our Random Forest model
    rf_model,

    # Use the training features
    X_train,

    # Use the training target values
    Y_train,

    # Divide the training data into 5 folds
    cv=5,

    # Use accuracy as the evaluation metric
    scoring='accuracy'
)

# Print the accuracy obtained from each fold
print("Cross-validation scores:", cv_scores)

# Calculate and print the average cross-validation accuracy
print("Mean CV Accuracy:", cv_scores.mean())
```

```python
# Import GridSearchCV for testing different Random Forest parameter combinations
from sklearn.model_selection import GridSearchCV

# Create a smaller parameter grid to make the search much faster
param_grid = {

    # Test three different numbers of trees
    'n_estimators': [100, 200],

    # Test three different maximum tree depths
    'max_depth': [3, 5, 7],

    # Test two different minimum sample requirements for splitting
    'min_samples_split': [2, 5],

    # Test two different minimum leaf sizes
    'min_samples_leaf': [1, 2],

    # Use the square-root feature selection strategy
    'max_features': ['sqrt']
}

# Create the Random Forest model that will be tuned
rf_for_grid = RandomForestClassifier(

    # Set a fixed random state so results are reproducible
    random_state=2
)

# Create the GridSearchCV object
grid_search = GridSearchCV(

    # Tell GridSearchCV to tune our Random Forest model
    estimator=rf_for_grid,

    # Provide the smaller parameter grid
    param_grid=param_grid,

    # Use 5-fold cross-validation
    cv=5,

    # Choose the model with the highest accuracy
    scoring='accuracy',

    # Use all available CPU cores
    n_jobs=-1
)

# Train the different Random Forest configurations
grid_search.fit(X_train, Y_train)

# Display the best parameter combination
print("Best Parameters:", grid_search.best_params_)

# Display the best average cross-validation accuracy
print("Best CV Accuracy:", grid_search.best_score_)
```

# **SVM-SUPPORT VECTOR MACHINE**

```python
# Import StandardScaler to standardize the numerical features
from sklearn.preprocessing import StandardScaler

# Import Pipeline so scaling and SVM are treated as one complete model
from sklearn.pipeline import Pipeline

# Import SVC, which is the Support Vector Classification algorithm
from sklearn.svm import SVC

# Import cross_val_score to evaluate the model using cross-validation
from sklearn.model_selection import cross_val_score

# Import the evaluation metrics we need
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python
# Create a pipeline containing both feature scaling and the SVM model
svm_model = Pipeline([

    # Standardize the features so they have comparable scales
    ('scaler', StandardScaler()),

    # Create the SVM classifier using the RBF kernel
    ('svm', SVC(kernel='rbf'))
])
```

```python
# Train the complete pipeline using the training data
svm_model.fit(X_train, Y_train)
```

```python
# Use the trained SVM pipeline to predict the training data
Y_train_prediction = svm_model.predict(X_train)

# Use the trained SVM pipeline to predict the unseen test data
Y_test_prediction = svm_model.predict(X_test)
```

```python
# Calculate the training accuracy
training_accuracy = accuracy_score(Y_train, Y_train_prediction)

# Calculate the training precision
training_precision = precision_score(Y_train, Y_train_prediction)

# Calculate the training recall
training_recall = recall_score(Y_train, Y_train_prediction)

# Calculate the training F1 score
training_f1 = f1_score(Y_train, Y_train_prediction)
```

```python
# Calculate the test accuracy
test_accuracy = accuracy_score(Y_test, Y_test_prediction)

# Calculate the test precision
test_precision = precision_score(Y_test, Y_test_prediction)

# Calculate the test recall
test_recall = recall_score(Y_test, Y_test_prediction)

# Calculate the test F1 score
test_f1 = f1_score(Y_test, Y_test_prediction)

# Calculate the confusion matrix
test_cm = confusion_matrix(Y_test, Y_test_prediction)
```

```python
# Print a heading for the SVM results
print("SVM Results")

# Print the training accuracy
print("Training Accuracy :", training_accuracy)

# Print the training precision
print("Training Precision:", training_precision)

# Print the training recall
print("Training Recall   :", training_recall)

# Print the training F1 score
print("Training F1 Score :", training_f1)

# Print the test accuracy
print("Test Accuracy     :", test_accuracy)

# Print the test precision
print("Test Precision    :", test_precision)

# Print the test recall
print("Test Recall       :", test_recall)

# Print the test F1 score
print("Test F1 Score     :", test_f1)

# Print the confusion matrix
print("Confusion Matrix:")

# Display the confusion matrix
print(test_cm)
```

# **LOGISTIC REGRESSION**

```python
# Import Pipeline so we can combine feature scaling and Logistic Regression
from sklearn.pipeline import Pipeline

# Import StandardScaler to standardize the input features
from sklearn.preprocessing import StandardScaler

# Import Logistic Regression as our next classification model
from sklearn.linear_model import LogisticRegression

# Create a pipeline that first scales the data and then trains Logistic Regression
logistic_pipeline = Pipeline([

    # Standardize the features before giving them to Logistic Regression
    ('scaler', StandardScaler()),

    # Create the Logistic Regression classifier
    ('logistic', LogisticRegression(max_iter=1000))
])
```

```python
# Create a parameter grid for tuning Logistic Regression
logistic_param_grid = {

    # Test different strengths of regularization
    'logistic__C': [0.01, 0.1, 1, 10, 100],

    # Test both L1 and L2 regularization
    'logistic__penalty': ['l1', 'l2'],

    # Use the solver that supports both L1 and L2 regularization
    'logistic__solver': ['liblinear']
}
```

```python
# Import GridSearchCV for finding the best Logistic Regression parameters
from sklearn.model_selection import GridSearchCV

# Create the GridSearchCV object for Logistic Regression
logistic_grid = GridSearchCV(

    # Use the scaling and Logistic Regression pipeline
    estimator=logistic_pipeline,

    # Give GridSearchCV the Logistic Regression parameter combinations
    param_grid=logistic_param_grid,

    # Use 5-fold cross-validation
    cv=5,

    # Select the model with the highest average accuracy
    scoring='accuracy',

    # Use all available CPU cores
    n_jobs=-1
)
```

```python
# Train all Logistic Regression configurations using 5-fold cross-validation
logistic_grid.fit(X_train, Y_train)
```

```python
# Print the best Logistic Regression parameters
print("Best Parameters:", logistic_grid.best_params_)

# Print the best average cross-validation accuracy
print("Best CV Accuracy:", logistic_grid.best_score_)
```

```python
# Get the best Logistic Regression pipeline found by GridSearchCV
best_logistic_model = logistic_grid.best_estimator_

# Use the best Logistic Regression model to predict the test data
logistic_test_prediction = best_logistic_model.predict(X_test)

# Calculate the test accuracy
logistic_test_accuracy = accuracy_score(Y_test, logistic_test_prediction)

# Calculate the test precision
logistic_test_precision = precision_score(Y_test, logistic_test_prediction)

# Calculate the test recall
logistic_test_recall = recall_score(Y_test, logistic_test_prediction)

# Calculate the test F1 score
logistic_test_f1 = f1_score(Y_test, logistic_test_prediction)

# Calculate the confusion matrix
logistic_cm = confusion_matrix(Y_test, logistic_test_prediction)

# Print the test accuracy
print("Test Accuracy :", logistic_test_accuracy)

# Print the test precision
print("Test Precision:", logistic_test_precision)

# Print the test recall
print("Test Recall   :", logistic_test_recall)

# Print the test F1 score
print("Test F1 Score :", logistic_test_f1)

# Print the confusion matrix
print("Confusion Matrix:")

# Display the confusion matrix
print(logistic_cm)
```
